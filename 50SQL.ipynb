```python
import pandas as pd
import sqlite3

customers = pd.DataFrame({
    'CustomerID':[101,102,103,104,105],
    'CustomerName':['Ravi','Neha','Rahul','Priya','Kiran'],
    'City':['Hyderabad','Vijayawada','Chennai','Hyderabad','Bangalore'],
    'Gender':['Male','Female','Male','Female','Male']
})


orders = pd.DataFrame({
    'OrderID':[1,2,3,4,5],
    'CustomerID':[101,102,103,101,104],
    'ProductName':['Laptop','Mobile','Keyboard','Mouse','Monitor'],
    'Quantity':[1,2,3,2,1],
    'Price':[55000,18000,1500,800,12000]
})


conn = sqlite3.connect("shop.db")

customers.to_sql("Customers", conn, if_exists="replace", index=False)
orders.to_sql("Orders", conn, if_exists="replace", index=False)

query = "SELECT * FROM Customers"
print(pd.read_sql(query, conn))
```
